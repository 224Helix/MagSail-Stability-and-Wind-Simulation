### Goal Definition

The goal of this project is to investigate the utility of MagSail propulsion, as proposed by Zubrin's 2000 Final Report to NIAC, at planetary-range orbits in M-type star systems. Such observations could be critical, not only for close-range habitability surveys, but for investigations into the behavior of the stars, which are highly dynamic and turbulent, allowing for verification of MHD simulations and greater understanding of stellar plasmas.
The target of this investigation is to return first-order predictions for:
1. The performance of such a propulsion system within these conditions where an science probe could benefit from a fuel-less propulsion system;
2. The capability of MagSail-equipped scientific instruments to maintain attitude necessary for making any planetary, stellar, or environmental measurements while in orbit. This will be constrained by the maximum possible slew rate of a parameterized tether system;
3. The capability of a MagSail to generate the thrust necessary to accelerate a craft in such a stellar orbit for various orbital maneuevers typically performed during scientific missions by chemical rockets;
4. The necessary construction and engineering parameters for such a probe, including comparisons to current commercial, experimental, and near-future predictions,
5. The influence of parameters within the simulation environment, and construction parameters of the MagSail, on (1) and (2), within the ranges as bounded by the literature as the author understands;

# STELLAR ENVIRONMENT

Essential to this project is the construction of a reasonable environment in which the MagSail simulation will take place. For this, a wind tunnel axially aligned with the star, which is prototyped off of Proxima Centauri, with a cross-section of 2,000km by 2,000km and an axial depth of 10,000km. These dimensions were chosen in response to Zubrin (2000)'s estimations for the magnetopause radius of a near-future operational magsail at ~20km, which takes up ~0.0314% of the cross-sectional area and thus be subject to more realistic, larger-scale turbulent events. Furthermore, this investigation will assume an orbit that lays close to the equator, to avoid powerful bipolar jets.

Firstly, Kavanagh et al. (2021) assert that the orbit of Proxima Centauri b is beyond the Alfven surface during quiescent phases of the star's magnetic cycle. However, during more aggravated cycles, the planet can spend a considerable amont of orbit time within the Alfven surface. The solar Alfven surface has been found to expand asymmetrically up to 30% by Badman et. al (2025). In stellar wind physics, the average radius of the Alfvén surface scales with the open magnetic flux, which is derived from the surface magnetic field strength according to an empirical power-law relationship typically modeled as: Alfven Radius ∝ B^[m]. For most magnetized stellar wind models (such as the standard 2.5D/3D magnetohydrodynamic formulations by Finley, Matt, and Réville), the exponent [m] usually hovers around 0.3 to 0.4 (with [m] ~ 0.33 being a common baseline for a split-monopole or dipole-dominated wind structure). 

Proxima Centauri's global magnetic field strength oscillates by a factor of roughly 4 (from ~500 Gauss up to ~2,000 Gauss). Using the B^[m] scaling approximation, we get bounds of [1.516-1.741]. When this is compounded with the 30% deviation from asymmetry shown in Badman et al. (2025), it gives a final range of [1.9708-2.2633] Thus, the orbital distance of the probe in this simulation will be rounded up to 2.5x that of Prox Cen b, at ~169.6 [R_star] or 18,150,000 km. This is [R_probe]. This places the probe well outside of the range of the Alfven surface at its greatest, preventing the myriad of problems that a sub-Alfvenic orbit would pose, including possible noise fir radio-based transmission needed for meaningful investigation; bi-directional causal relationships where the MHD waves from the probe could return to the star, confounding possible measurements; and the likely engineering constraints of shielding a craft that orbits for long periods within this region.

As for the structure of the magnetic field, we will use the source-surface measurements from Garraffo, which place the mean at $600\,\mathrm{G}$ and the maximum at $1200\,\mathrm{G}$. Using these values, we will employ the Potential Field Source Surface (PFSS) model to estimate the magnetic field at our $R_{\mathrm{probe}}$. The spiral tightness factor measures how much the azimuthal curl of the field affects the magnetic-field geometry. It is defined as it is defined as

$$
\mathcal{T}_{\mathrm{spiral}}
=
\Omega\frac{R-R_{ss}}{v_w}
$$

where $\Omega$ is the stellar angular velocity. Given the rotational period of Prox Cen being $\sim 85$ days, we find that $\mathcal{T}_{\mathrm{spiral}} \approx 0.019$.

$$
B_r(R) = B_* \left(\frac{R_*}{R_{ss}}\right)^3 \left(\frac{R_{ss}}{R}\right)^2 = B_* \frac{R_*^3}{R_{ss} R^2}
$$
Plugging these assumptions into our two sets of B-field conditions from Garraffo et al. and employing the following:
1. R_{ss} is taken, as is convention, to be 2.5 [R_star] or 267,500 km

Through this, we obtain these figures:

Using

## Characteristic Length Check

In Zubrin's work, the standoff distance is derived from the pressure balance at the magnetopause and given as:

$$L_0 = \left( \frac{\mu_0 M^2}{8\pi^2 n m_i u^2} \right)^{1/6}$$

Looking beyond the geometric and physical factors, this distance is a function of:
 *M* - the magnetic dipole moment of the Sail, 
 *n*- the numerical density of the charges in the plasma, and
 *u* - the relative velocity of the Sail to the plasma

Computing $L_0$ with the given constants: <br>
 $μ_0 = 4π×10^{−7}$ T·m/A <br>
 $𝑚_𝑖 = 1.6726 × 10^{−27}$ kg (proton mass) 

Using Zubrin's figures for the operational sail's geometry, we obtain: <br>
$$M \approx 2.01 \times 10^{13} \text{ A}\cdot\text{m}^2 \quad (\text{or } \text{J/T})$$

Finally, the stellar wind of Proxima Centauri is expected to vary substantially due to the star's strong magnetic activity and coronal structure. MHD simulations by Garraffo et al. (2016) predict large spatial variations in wind density, velocity, and magnetic pressure at the orbit of Proxima b. To account for this variability, this work models the stellar wind using a parameter space rather than a single steady-state value. Thus, we will adopt a range of particle densities (100 − 300 cm^−3) and velocities (300−800 km s^−1) encompassing values derived from Ly-α-based mass-loss estimates (Wood et al. 2001) and global MHD simulations of the Proxima system (Garraffo et al. 2016).

To find L_0:
We can separate the constants from the variable wind factors:

$$C = \frac{\mu_0 M^2}{8\pi^2 m_i}$$

$$\mu_0 M^2 = (4\pi \times 10^{-7}) \times (2.01 \times 10^{13})^2 \approx 5.07694 \times 10^{20}$$

$$8\pi^2 m_i = 8\pi^2 \times (1.6726 \times 10^{-27}) \approx 1.32063 \times 10^{-25}$$

$$C \approx \frac{5.07694 \times 10^{20}}{1.32063 \times 10^{-25}} \approx 3.84428 \times 10^{45} \text{ m}^6/(\text{m}^3 \cdot \text{s}^{-2})$$

Then, the standoff distance becomes:

$$L_0 = \left( \frac{C}{n u^2} \right)^{1/6}$$

Plugging in the minimum and maximum dynamic pressure ($nu^2$), we find that
$L_0$ ranges between $16.5 \text{ km}$ and $27.4 \text{ km}$ ($16,478\text{ m} - 27,442\text{ m}$).

This characteristic length, when compared to the gyroradius, gives a unitless scale factor between ~ 0.2239 and 0.0675.

## JOURNAL ENTRY - 8.7.2026

The above characteristic length derivation justifies a fluid approach for most of the region surrounding the MagSail. However, there are smaller, yet still relevant length scales at the boundary layer. Thus, rather than assuming full fidelity, which would be non-physical, the model will proceed with a method inspired by Siscoe's *Solar System Magnetohydrodynamics*. In the work, the author utilizes the generalized Ohm's law defined as:

$$\vec{E} + \vec{V} \times \vec{B} = \eta\vec{J} + \frac{1}{ne}\left(\vec{J} \times \vec{B}\right) - \frac{1}{ne}\nabla \cdot \mathbf{P}_e + \frac{m_e}{ne^2}\left[\frac{\partial \vec{J}}{\partial t} + \nabla \cdot (\vec{V}\vec{J} + \vec{J}\vec{V})\right]$$

And compares *across space* the magnitude of the motional term $V \times B$ with 4 other terms ($N_1 - N_4$) corresponding to the contributions of physical behaviors:
1. Ohmic Resistance Term ($\eta\vec{J}$): Represents momentum loss due to collisions between particles (resistivity $\eta$ and current density $\vec{J}$). <br>  
2. Hall Effect Term ($\frac{1}{ne}\vec{J} \times \vec{B}$): Represents the magnetic force acting on the charge carriers, accounting for the decoupling of electron and ion motion. <br>
3. Electron Pressure Gradient Term ($\frac{1}{ne}\nabla \cdot \mathbf{P}_e$): Represents the ambipolar electric field created when electron pressure gradients drive electron diffusion relative to ions. <br>
4. Electron Inertia Term ($\frac{m_e}{ne^2}\frac{\mathrm{d}\vec{J}}{\mathrm{d}t}$): Represents  the finite mass ($m_e$) and time lag of electrons responding to changing forces.  

Thus, the project will utilize an **Adaptive Mesh Refinement** system, where spaces with a greater gap in magnitude between the motional term and these terms are rendered in lower resolution, in order to efficiently utilize compute while maintaining kinetic and hybrid effects.



## METHODOLOGY

The AMR will be developed by comparing the motional term to $N_1 through N_4$ across space. Firstly, we need to select a distance at which the stellar B-field dominates the geometry. This will allow us to apply a fluid approach to the region before this boundary.

Once literature review is concluded for finding this boundary, the fluid regime will be efficiently simulated by generating a spectral field that is advected through space. Transient events like flares, clumps, and shocks will be advected through this field.
